```sql
create table if not exists gizmobox.gold.customer_addresses
as
select c.customer_id as customer_id,
    c.customer_name as customer_name,
    c.date_of_birth as date_of_birth,
    c.email as email,
    c.member_since as member_since,
    c.telephone as telephone,
    a.shipping_address_line_1 as shipping_address_line_1,
    a.shipping_city as shipping_city,
    a.shipping_state as shipping_state,
    a.shipping_postcode as shipping_postcode,
    a.billing_address_line_1 as billing_address_line_1,
    a.billing_city as billing_city,
    a.billing_state as billing_state,
    a.billing_postcode as billing_postcode
 from gizmobox.silver.customers c
   inner join gizmobox.silver.addresses a
     on c.customer_id = a.customer_id

-- DBTITLE 1
```
