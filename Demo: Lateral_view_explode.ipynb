```sql
select schema_of_json(order_info) from json_with_array
```

```sql
create or replace temp view json_with_array_struct as
with cte_name
as
(
select order_id,
 from_json(order_info, 'STRUCT<customer: STRING, items: ARRAY<STRUCT<price: DOUBLE, product: STRING, qty: BIGINT>>>') as order_info_struct
 from json_with_array
)
select order_id,
      order_info_struct.customer as customer,
      explode(order_info_struct.items) as item
from cte_name
```

```sql
select * from json_with_array_struct
```

```sql
select order_id, customer, 
       item.price as price,
       item.product as product,
       item.qty as qty 
from json_with_array_struct
```
